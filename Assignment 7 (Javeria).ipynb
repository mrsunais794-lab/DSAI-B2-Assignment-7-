# Question 1: Employee Data Bank (Using OOP)


```python
class Employee:
    def __init__(self, name, number, address, salary, post,
                 year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print("=" * 45)
        print(f"  Employee Detail Information")
        print("=" * 45)
        print(f"Name                : {self.name}")
        print(f"Employee Number     : {self.number}")
        print(f"Address             : {self.address}")
        print(f"Salary              : {self.salary}")
        print(f"Post                : {self.post}")
        print(f"Year of Joining     : {self.year_of_joining}")
        print(f"Education           : {self.education}")
        print(f"Year of Experience  : {self.year_of_experience}")
        print("=" * 45)

    def update_year_of_experience(self, years=1):
        self.year_of_experience += years
        print(f"{self.name}'s experience updated to "
              f"{self.year_of_experience} year(s).")

    def increase_salary(self, amount):
        self.salary += amount
        print(f"{self.name}'s salary increased by {amount}. "
              f"New salary: {self.salary}")



print("\n########## Instance: employee101 ##########")
employee101 = Employee("Ali Khan", 101, "Lahore", 60000,
                       "Software Engineer", 2020, "BS Computer Science", 3)

print("Individual attributes of employee101:")
print("Name   :", employee101.name)
print("Number :", employee101.number)
print("Salary :", employee101.salary)
print("Post   :", employee101.post)

employee101.describe_employee()
employee101.increase_salary(5000)
employee101.update_year_of_experience()
employee101.describe_employee()

print("\n########## Ten Employees ##########")

employees = [
    Employee("Ali Khan",    101, "Lahore",      60000, "Software Engineer", 2020, "BS CS",        3),
    Employee("Sara Ahmed",  102, "Karachi",     75000, "Data Analyst",      2019, "MS Data Sci",  5),
    Employee("Bilal Raza",  103, "Islamabad",   50000, "Junior Developer",  2021, "BS IT",        2),
    Employee("Hina Malik",  104, "Faisalabad",  90000, "Project Manager",   2017, "MBA",          7),
    Employee("Usman Tariq", 105, "Multan",      45000, "QA Engineer",       2022, "BS SE",        1),
    Employee("Ayesha Noor", 106, "Peshawar",    80000, "HR Manager",        2018, "MHRM",         6),
    Employee("Hamza Sheikh",107, "Quetta",      55000, "Network Admin",     2020, "BS Telecom",   3),
    Employee("Zainab Iqbal",108, "Sialkot",     70000, "UI/UX Designer",    2019, "BS Design",    4),
    Employee("Fahad Anwar", 109, "Hyderabad",   65000, "Backend Developer", 2021, "BS CS",        2),
    Employee("Maryam Shah", 110, "Gujranwala",  95000, "AI Engineer",       2016, "MS AI",        8),
]

for emp in employees:
    emp.describe_employee()

print("\n########## Updates ##########")
for emp in employees:
    emp.increase_salary(3000)
    emp.update_year_of_experience()

print("\n########## Updated Details (first 3 shown) ##########")
for emp in employees[:3]:
    emp.describe_employee()
```

# Question 2: Same Task Without Classes / OOP


```python
def create_employee(name, number, address, salary, post,
                    year_of_joining, education, year_of_experience):
    """Acts like __init__ - returns an employee dictionary."""
    return {
        "name": name,
        "number": number,
        "address": address,
        "salary": salary,
        "post": post,
        "year_of_joining": year_of_joining,
        "education": education,
        "year_of_experience": year_of_experience,
    }


def describe_employee(emp):
    print("=" * 45)
    print("  Employee Detail Information")
    print("=" * 45)
    for key, value in emp.items():
        label = key.replace("_", " ").title()
        print(f"{label:<20}: {value}")
    print("=" * 45)


def update_year_of_experience(emp, years=1):
    emp["year_of_experience"] += years
    print(f"{emp['name']}'s experience updated to "
          f"{emp['year_of_experience']} year(s).")


def increase_salary(emp, amount):
    emp["salary"] += amount
    print(f"{emp['name']}'s salary increased by {amount}. "
          f"New salary: {emp['salary']}")


print("\n########## Instance: employee101 ##########")
employee101 = create_employee("Ali Khan", 101, "Lahore", 60000,
                              "Software Engineer", 2020,
                              "BS Computer Science", 3)

print("Individual attributes of employee101:")
print("Name   :", employee101["name"])
print("Number :", employee101["number"])
print("Salary :", employee101["salary"])
print("Post   :", employee101["post"])

describe_employee(employee101)
increase_salary(employee101, 5000)
update_year_of_experience(employee101)
describe_employee(employee101)


print("\n########## Ten Employees ##########")

employees = [
    create_employee("Ali Khan",    101, "Lahore",      60000, "Software Engineer", 2020, "BS CS",        3),
    create_employee("Sara Ahmed",  102, "Karachi",     75000, "Data Analyst",      2019, "MS Data Sci",  5),
    create_employee("Bilal Raza",  103, "Islamabad",   50000, "Junior Developer",  2021, "BS IT",        2),
    create_employee("Hina Malik",  104, "Faisalabad",  90000, "Project Manager",   2017, "MBA",          7),
    create_employee("Usman Tariq", 105, "Multan",      45000, "QA Engineer",       2022, "BS SE",        1),
    create_employee("Ayesha Noor", 106, "Peshawar",    80000, "HR Manager",        2018, "MHRM",         6),
    create_employee("Hamza Sheikh",107, "Quetta",      55000, "Network Admin",     2020, "BS Telecom",   3),
    create_employee("Zainab Iqbal",108, "Sialkot",     70000, "UI/UX Designer",    2019, "BS Design",    4),
    create_employee("Fahad Anwar", 109, "Hyderabad",   65000, "Backend Developer", 2021, "BS CS",        2),
    create_employee("Maryam Shah", 110, "Gujranwala",  95000, "AI Engineer",       2016, "MS AI",        8),
]

for emp in employees:
    describe_employee(emp)

print("\n########## Updates ##########")
for emp in employees:
    increase_salary(emp, 3000)
    update_year_of_experience(emp)

print("\n########## Updated Details (first 3 shown) ##########")
for emp in employees[:3]:
    describe_employee(emp)
```
